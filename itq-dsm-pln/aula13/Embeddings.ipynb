```python
from gensim.models import KeyedVectors
```

```python
import numpy as np
```

```python
modelo = KeyedVectors.load_word2vec_format("D:/git/dados/nlp/skip_s100/skip_s100.txt")
```

```python
def hipotenusa( vetor ):
    soma = 0 
    for valor in vetor:
        soma += (valor ** 2)
    return np.sqrt( soma )
```

```python
estudante = modelo["estudante"]
aluno= modelo["aluno"]

hipo_estudante = hipotenusa( estudante )
hipo_aluno = hipotenusa( aluno )

# similaridade = np.dot(aluno, estudante) / hipo_aluno * hipo_estudante
similaridade = np.dot(aluno / hipo_aluno, estudante / hipo_estudante)
print("Similaridade: ", similaridade)
```
