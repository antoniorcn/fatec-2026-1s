```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
import random
```

```python
X1 = np.array([150000, 50000, 250000, 300000, 450000, 1500000])
X2 = np.array([35, 10, 40, 20, 50, 100])

X1_min, X1_max = X1.min(), X1.max()
X1_norm = (X1 - X1_min) / (X1_max - X1_min)

X2_min, X2_max = X2.min(), X2.max()
X2_norm = (X2 - X2_min) / (X2_max - X2_min)

# 1 - Comprar e 0 Não comprar
Y = np.array([1, 0, 1, 0, 1, 0])
```

```python
X1_norm
```

```python
W0 = random.randint(-100, 100)
W1 = random.randint(-100, 100)
W2 = random.randint(-100, 100)
resultados = [0, 0, 0, 0, 0, 0]
learning_rate = 1
for j in range(100000):
    for i in range(len(X1_norm)):
        somatoria = (1 * W0) + (X1_norm[i] * W1) + (X2_norm[i] * W2)
        resultado = 1 if somatoria > 0 else 0
        resultados[i] = resultado
        erro = Y[i] - resultado
        W1 = W1 + (learning_rate * X1_norm[i] * erro)
        W2 = W2 + (learning_rate * X2_norm[i] * erro)
print(f"Resultados obtidos: {resultados}\t reais: {Y}")
```

```python
valor = 300000
valor_norm = (valor - X1_min) / (X1_max - X1_min)
m2 = 70
m2_norm = (m2 - X2_min) / (X2_max - X2_min)

somatoria = (1 * W0) + (valor_norm * W1) + (m2_norm * W2)
resultado = 1 if somatoria > 0 else 0
print("Voce deve comprar este imovel ==>", resultado)
```

```python
cores = []
for item in resultados:
    cores.append( "red" if item <= 0 else "green" )

plt.scatter(X2, X1, color=cores)
```
