```python
import torch
from torch.nn import Linear, MSELoss
from torch.optim import SGD
```

```python
casas_m2 = torch.tensor([[10], [13], [15], [17], [24], [28], [30], [45], [52], [64], 
                         [70], [80], [90], [100], [110], [120], [160]], 
                        dtype = torch.float32)
casas_valores = torch.tensor([[150000], [180000], [185000], [190000], [230000], [250000], [280000], [430000], [500000], 
                              [620000], [710000], [830000], [920000], [1050000], [1100000], [900000], [1300000]], 
                             dtype = torch.float32)
```

```python
X = casas_m2
Y = casas_valores
```

```python
modelo = Linear(in_features=1, out_features=1)
```

```python
for p in modelo.parameters():
    print(p)
```

```python
# Mean Squared Error
criterio_perda = MSELoss()
```

```python
optimizador = SGD(modelo.parameters(), lr=0.0001)
```

```python
for idx, epoca in enumerate(range(100)):
    y_hat = modelo(X)
    loss = criterio_perda(y_hat, Y)
    print(f"Epoca: {idx + 1}  Perda: {loss} ")
    optimizador.zero_grad()
    loss.backward()
    optimizador.step()
```

```python
with torch.no_grad():
    m2 = torch.tensor([[85.0]])
    valor_aproximado = modelo( m2 )
    print("Valor aproximado para: 75.0 ==> ", valor_aproximado.item())
```
