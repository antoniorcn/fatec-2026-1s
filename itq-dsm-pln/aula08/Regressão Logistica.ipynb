```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
# Tabela AND
#  Feature 1    |    Feature 2    | Label 
#     1         |       1         |   1
#     0         |       1         |   0
#     1         |       0         |   0
#     0         |       0         |   0

X = torch.tensor([
    [1.0, 1.0],
    [0.0, 1.0],
    [1.0, 0.0],
    [0.0, 0.0],
])

Y = torch.tensor([ 
    [1.0], 
    [0.0], 
    [0.0], 
    [0.0] 
])
```

```python
modelo = nn.Sequential(
    nn.Linear(in_features = 2, out_features = 1),
    nn.Sigmoid()
)
```

```python
criterio = nn.BCELoss()   # Binary Cross Entropy Loss
otimizador = optim.SGD(modelo.parameters(), lr = 1)
```

```python
for idx, epoca in enumerate(range(100)):
    Y_hat = modelo( X )
    loss = criterio( Y_hat, Y )

    otimizador.zero_grad()
    loss.backward()
    otimizador.step()

    print(f"Epoca: {idx + 1}\tLoss: {loss}")
```

```python
with torch.no_grad():
    resultado = modelo( torch.tensor([[0.0, 0.0]]) )
    print(f"Resultado: {resultado.item() >= 0.5}")
```
