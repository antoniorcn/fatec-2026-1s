```python
import pandas as pd
```

```python
import torch
from torch import nn, optim
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
df = pd.read_csv("D:/git/dados/nlp/news_sentiment_analysis.csv", encoding="utf-8")
```

```python
df = df[["Description", "Sentiment"]]
```

```python
df["sentiment_number"] = df["Sentiment"].map( {"positive": 1, "negative": 0, "neutral": 0} )
```

```python
Y = torch.tensor(df["sentiment_number"], dtype=torch.float32)
Y = torch.reshape( Y, (-1, 1) )
Y
```

```python
vetorizador = CountVectorizer(max_features = 1000)
```

```python
X = vetorizador.fit_transform( df["Description"] )
X = torch.tensor(X.toarray(), dtype=torch.float32)
X
```

```python
vocabulario = vetorizador.get_feature_names_out()
```

```python
modelo = nn.Sequential(
    nn.Linear(in_features = 1000, out_features = 1), 
    nn.Sigmoid()
)
otimizador = optim.SGD( modelo.parameters(), lr=1)
criterio = nn.BCELoss() # Binary Cross Entropy Perda
```

```python
for epoca in range(1000):
    y_hat = modelo( X ) 
    loss = criterio( y_hat, Y )
    print(f"Loss: {loss}")
    otimizador.zero_grad()
    loss.backward()
    otimizador.step()
```

```python
vetorizador_teste = CountVectorizer( vocabulary = vocabulario )
```

```python
X_teste = vetorizador_teste.fit_transform( [texto] )
X_teste = torch.tensor( [X_teste.toarray()], dtype=torch.float32 )
X_teste
```

```python
Y_hat_teste = modelo( X_teste )
Y_hat_teste.item()
```
