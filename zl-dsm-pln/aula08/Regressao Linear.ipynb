```python
import torch
from torch import nn
from torch import optim
```

```python
casas_m2 = torch.tensor([[10], [13], [15], [17], [24], [28], [30], [45], [52], [64], 
                         [70], [80], [90], [100], [110], [120], [160]], 
                        dtype = torch.float32)
casas_valores = torch.tensor([[150000], [180000], [185000], [190000], [230000], [250000], [280000], [430000], [500000], 
                              [620000], [710000], [830000], [920000], [1050000], [1100000], [900000], [1300000]], 
                             dtype = torch.float32)
```

```python
casas_m2
```

```python
casas_valores
```

```python
modelo = nn.Linear(in_features = 1, out_features = 1)
criterio = nn.MSELoss() # Mean Squared Error
otimizador = optim.SGD(modelo.parameters(), lr = 0.00001) # Stocastic Gradient Descent
```

```python
for epoca in range(100):
    y_hat = modelo( casas_m2 )
    loss = criterio( y_hat, casas_valores )
    print(f"Epoca: {epoca}\tLoss: {loss}")
    otimizador.zero_grad()
    loss.backward()
    otimizador.step()
```
