```python
import torch
```

```python
n1 = torch.tensor( 15, dtype = torch.float32 ) 
n1
```

```python
print(f"Shape: {n1.shape}") 
print(f"Dimensoes: {n1.ndim}") 
print(f"Type: {n1.dtype}")
print(f"Valor: {n1.item()}")
```

```python
v1 = torch.tensor( [10, 15, 20] )
```

```python
print(f"Shape: {v1.shape}") 
print(f"Dimensoes: {v1.ndim}") 
print(f"Type: {v1.dtype}")
print(f"Valor: {v1[0].item()}")
```

```python
m1 = torch.randn( (5, 3, 2) )
print(f"Shape: {m1.shape}") 
print(f"Dimensoes: {m1.ndim}") 
print(f"Type: {m1.dtype}")
print(f"Valor: {m1[0, 0, 0].item()}")
```
