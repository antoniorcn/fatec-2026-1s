```python
import matplotlib.pyplot as plt 
```

```python
casas_m2 = np.array([10, 13, 15, 17, 24, 28, 30, 45, 52, 64, 70, 80, 90, 100, 110, 120, 160])
casas_valores = np.array([150000, 180000, 185000, 190000, 230000, 250000, 280000, 430000, 500000, 620000, 710000, 830000, 920000, 1050000, 1100000, 900000, 1300000])
```

```python
casas_teste_m2 = np.array([150, 160, 155, 140, 170, 165, 152])
casas_teste_valores = np.array([300000, 350000, 320000, 280000, 400000, 370000, 310000])
```

```python
X = casas_m2
Y = casas_valores
```

```python
w = 2300
b = 0

def calculo_y_hat( x, w, b):
    return x * w + b

def custo( X, Y, w, b ):
    soma_erros = 0
    for idx, x in enumerate(X):
        y_hat = calculo_y_hat( x, w, b )
        y = Y[idx]
        erro = (y_hat - y) ** 2
        soma_erros = soma_erros + erro
    valor_custo = soma_erros / (2 * len(X))
    return valor_custo


Y_hat = calculo_y_hat( X, w, b )


valor_custo = custo( X, Y, w, b )
print(f"W:{w}\tB:{b}\tCusto:{valor_custo}")

plt.scatter( X, Y, label="Valores Reais")
plt.plot( X, Y_hat, color="red", marker="o", label="Valores Estimados" )
plt.legend()
plt.show()
```

```python
w = 1000
b = 500
alpha = 0.00001

def funcao_custo( X, Y, w, b, alpha ):
    soma_erros_dw = 0
    soma_erros_db = 0
    soma_erros_squared = 0
    m = len( X ) 
    for idx, x in enumerate( X ):
        y_hat = w * x + b
        y = Y[idx]
        erro = y_hat - y
        erro_squared = erro ** 2
        soma_erros_dw = soma_erros_dw + (erro * x)
        soma_erros_db = soma_erros_db + erro
        soma_erros_squared = soma_erros_squared + erro_squared
    dw = soma_erros_dw / m
    db = soma_erros_db / m
    novo_w = w - alpha * dw
    novo_b = b - alpha * db
    custo = soma_erros_squared / ( 2 * m )
    return custo, novo_w, novo_b
```

```python
for epoca in range(100):
    custo, w, b = funcao_custo(X, Y, w, b, alpha)
    print(f"W:{w}\tB:{b}\tCusto:{custo}")
```

```python
Y_hat = X * w + b
plt.scatter( X, Y, label="Valores Reais")
plt.plot( X, Y_hat, color="red", marker="o", label="Valores Estimados" )
plt.legend()
plt.show()
```
