```python
lousa = modelo["lousa"]
soma_lousa = 0
for valor in lousa:
    soma_lousa += (valor ** 2)
hipo_lousa = math.sqrt( soma_lousa )
hipo_lousa
```

```python
giz = modelo["giz"]
soma_giz = 0
for valor in giz:
    soma_giz += (valor ** 2)
hipo_giz = math.sqrt( soma_giz )
hipo_giz
```

```python
lousa_norm = lousa / hipo_lousa
giz_norm = giz / hipo_giz
```

```python
np.dot(lousa, giz) / (hipo_lousa * hipo_giz)
```
