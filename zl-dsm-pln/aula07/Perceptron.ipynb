```python
import numpy as np
import random
```

```python
VALORES = np.array([150000, 50000, 250000,300000, 450000, 15000000])
M2 = np.array([35, 10, 40, 20, 50, 100])
Y = np.array([1, 0, 1, 0, 1, 0])

VALORES_DIF = VALORES.max() - VALORES.min()
VALORES_NORM = (VALORES - VALORES.min()) / VALORES_DIF

M2_DIF = M2.max() - M2.min()
M2_NORM = (M2 - M2.min()) / M2_DIF

X1 = VALORES
X2 = M2
print(f"X1: {X1}")
print(f"X2: {X2}")
X1 = VALORES_NORM
X2 = M2_NORM
print(f"Normalizados  X1: {X1}")
print(f"Normalizados  X2: {X2}")
```

```python
W0 = random.randint(-10, 10)
W1 = random.randint(-100, 100)
W2 = random.randint(-100, 100)
```

```python
def mostrar_resultado():
    print(f"W1: {W1}\tW2:  {W2}")
    print("Y_Hat:\t[", end="")
    for i in range(len(X1)):
        somatoria = W0 + X1[i] * W1 + X2[i] * W2
        resultado = 1 if somatoria > 0 else 0
        print(resultado, end=" ", sep="")
    print(f"]\nY:\t{Y}")
```

```python
learning_rate = 1
mostrar_resultado()
for epoca in range(100000):
    for i in range(len(X1)):
        somatoria = W0 + (X1[i] * W1) + (X2[i] * W2)
        # threshold
        resultado = 1 if somatoria > 0 else 0
        erro = Y[i] - resultado
        W1 = W1 + (learning_rate * X1[i] * erro)
        W2 = W2 + (learning_rate * X2[i] * erro)
mostrar_resultado()
```
